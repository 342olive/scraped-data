```python
import numpy as np

np.random.seed(0)

X=[[1,2,3,2.5],
   [2.0, 5.0, -1.0, 2.0],
   [-1.5 2.7, 3.3, 0.8]]


Class Layer_Dense:
 def__init__(self,n_inputs,n_neurons):
  self.weights=np.random.randn(n_inputs,n_neurons)
  self.biases=np.zeroes(1,n_neurons)# the shape of baises matrix is decided as per numpy broadcasting rule
 def forward(self,inputs):
   self.output=(inputs*self.weights)+self.biases

Layer1=Layer_Dense(3,4)
Layer2=Layer_Dense(3,2) # layer 1 and 2 are dictioanry objects that are empty, it only has keys


Layer1.forward(X)#I am calling a method using this object so result will be stored in this object only
print(Layer1.output)
Layer2.forward(Layer1.output)# Layer1.output is used to retrieve the value of output key in the Layer1 dictionary object
print(Layer2.output) # retrieving the value of output key from layer 2 object , but here we are referring to the output key of Layer2 and not the output key of layer1
```
