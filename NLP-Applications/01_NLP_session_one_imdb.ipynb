```python
if __name__=="__main__":
  # read the training data
  df= pd.read_csv("/content/drive/MyDrive/NLP practice sessions/IMDB Dataset.csv")

  # note we assigned the changes values to same column
  df.sentiment= df.sentiment.apply(lambda x: 1 if x =="positive" else 0)

  # we create a new column and assign it back to the same dataframe
  df["kfold"]= -1

  #randomise the rows basically shuffle
  df= df.sample(frac=1).reset_index(drop=True)

  #initialise the kfold from model_selection module
  kf= model_selection.StratifiedKFold(n_splits=5)

  # Assuming 'y' is the target variable which is sentiment
  y = df.sentiment.values

  for f,(t_,v_) in enumerate(kf.split(X=df,y=y)):
    df.loc[v_,'kfold']=f # here f on right side means that we are assigning all the indices of v_ at once and updating the value of kfold by f, note we could have used nay column but we use kfold to replace its value because we dont want to lose data from any other column
     #note: use iloc instead of loc, loc is a bug
  # we go over the folds created
  # here using iloc we mean to say find indexes with these rows and replace their value with f
  for fold_ in range(5):
    # temporary dataframes for train and test
    train_df=df[df.kfold!=fold_].reset_index(drop=True)
    test_df=df[df.kfold==fold_].reset_index(drop=True) # we want to use a for loop to pick on set and create test out of it , rest all the sets shoulb become train_df which is why we use = condition

    train_path = f"/content/drive/MyDrive/NLP practice sessions/train_fold{fold_}.csv"
    test_path = f"/content/drive/MyDrive/NLP practice sessions/test_fold{fold_}.csv"

    train_df.to_csv(train_path, index=False)
    test_df.to_csv(test_path, index=False)

    print(f"Saved: {train_path} and {test_path}")


    # Load the punkt tokenizer data
    nltk.data.load('tokenizers/punkt/english.pickle')

  # initialize as well as call the CountVectorizer with NLTK's word_tokenize
  # intialize it as in create instance of a class and named it count_vec
    count_vec= CountVectorizer(
              tokenizer=word_tokenize,
    )

  # fit count_vec on training data reviews
    count_vec.fit(train_df.review)

  # transform training and vlaidation data
    xtrain=count_vec.transform(train_df.review)
    xtest=count_vec.transform(test_df.review) # Corrected typo from transform_df.review to test_df.review

  # initialize logistic regression model
    model = linear_model.LogisticRegression(max_iter=1000)

  # fit the model on training data reviews and sentiment
    model.fit(xtrain,train_df.sentiment)

  # make predictions on test data
  # threshold for predictions is 0.5
    preds = model.predict(xtest)

  # calculate accuracy
    accuracy = metrics.accuracy_score(test_df.sentiment, preds)

    print(f"Fold {fold_} Accuracy: {accuracy}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.tokenize import word_tokenize

# create a corpus of sentences
corpus = [ "hello, how are you?",
    "im getting bored at home. And you? What do you think?",
    "did you know about counts",
    "let's see if this works!",
    "YES!!!!"
]

# initialize TfidfVectorizer with word_tokenize from nltk
# as the tokenizer
tfv = TfidfVectorizer(tokenizer=word_tokenize, token_pattern=None)

# fit the vectorizer on corpus
tfv.fit(corpus)

corpus_transformed = tfv.transform(corpus)
print(corpus_transformed)
```
